## Data Columns Info

id 
Unique Identifier of the record

case
The Chicago Police Department RD Number (Records Division Number), which is unique to the incident

date
Date when the incident occurred. this is sometimes a best estimate.

block
The partially redacted address where the incident occurred, placing it on the same block as the actual address.

iucr
The Illinois Unifrom Crime Reporting code. This is directly linked to the Primary Type and Description. See the list of IUCR codes at https://data.cityofchicago.org/d/c7ck-438e.

primary_type
The primary description of the IUCR code.

description
The secondary description of the IUCR code, a subcategory of the primary description.

location_description
Description of the location where the incident occurred.

arrest
Indicates whether an arrest was made.

domestic
Indicates whether the incident was domestic-related as defined by the Illinois Domestic Violence Act.

beat
Indicates the beat where the incident occurred. A beat is the smallest police geographic area – each beat has a dedicated police beat car. Three to five beats make up a police sector, and three sectors make up a police district. The Chicago Police Department has 22 police districts. See the beats at https://data.cityofchicago.org/d/aerh-rz74.


district
Indicates the police district where the incident occurred. See the districts at https://data.cityofchicago.org/d/fthy-xz3r.

ward
The ward (City Council district) where the incident occurred. See the wards at https://data.cityofchicago.org/d/sp34-6z76.

community_area
Indicates the community area where the incident occurred. Chicago has 77 community areas. See the community areas at https://data.cityofchicago.org/d/cauq-8yn6.

fbi_code
Indicates the crime classification as outlined in the FBI's National Incident-Based Reporting System (NIBRS). See the Chicago Police Department listing of these classifications at http://gis.chicagopolice.org/clearmap_crime_sums/crime_types.html.

x_coordinate
The x coordinate of the location where the incident occurred in State Plane Illinois East NAD 1983 projection. This location is shifted from the actual location for partial redaction but falls on the same block.

y_coordinate
The y coordinate of the location where the incident occurred in State Plane Illinois East NAD 1983 projection. This location is shifted from the actual location for partial redaction but falls on the same block.

year
Year the incident occurred.

updated_on
Date and time the record was last updated.

latitude
The latitude of the location where the incident occurred. This location is shifted from the actual location for partial redaction but falls on the same block.

longitude
The longitude of the location where the incident occurred. This location is shifted from the actual location for partial redaction but falls on the same block.

location
The location where the incident occurred in a format that allows for creation of maps and other geographic

```python
df = spark.read.csv("dbfs:/Volumes/workspace/chicago_schema/chicago_volume/chicago_crimes_2015_2025.csv", header=True, inferSchema=True)
```

```python
# converting column names to lowercase for consistency
df = df.toDF(*[c.lower() for c in df.columns])

print(f"{df.count()} rows, {len(df.columns)} columns") # num rows and cols
display(df)
```

### Basic Data Cleaning
Removing nulls, duplicates, changing variable types etc.

```python
# Look at number of null values for all fields
from pyspark.sql.functions import col, sum

df.select([sum(col(c).isNull().cast("int")).alias(c) for c in df.columns]).show()
```

```python
# 2.5M rows in the data, approx 40k nulls for latitude, longitude and location.
# will just drop NA as have many data already

df_clean = df.dropna()
df_clean.count()
```

```python
from pyspark.sql.functions import count

total_rows = df_clean.count()
distinct_rows = df_clean.distinct().count()
duplicate_count = total_rows - distinct_rows

print(f"Number of duplicate rows: {duplicate_count}")
# no duplicates found
```

```python
# printing schema to examine variable types and see if any need to be changed
df_clean.printSchema()
```

```python
from pyspark.sql.functions import to_timestamp
from pyspark.sql.functions import date_format

# changing date cols to timestamp
df_clean = df_clean.withColumn("date", to_timestamp("date", "MM/dd/yyyy hh:mm:ss a")) \
                   .withColumn("updated on", to_timestamp("updated on", "MM/dd/yyyy hh:mm:ss a"))
# formatting that removes trailing zeros
df_clean = df_clean.withColumn("date", date_format("date", "dd/MM/yyyy HH:mm")) \
                   .withColumn("updated on", date_format("updated on", "dd/MM/yyyy HH:mm"))

# changing these integer variables to strings
df_clean = df_clean.withColumn("id", col("id").cast("string")) \
                   .withColumn("beat", col("beat").cast("string")) \
                   .withColumn("district", col("district").cast("string")) \
                   .withColumn("ward", col("ward").cast("string")) \
                   .withColumn("community area", col("community area").cast("string")) 
```

```python
df_clean.printSchema()
```

```python
# capitalising all 'block' entries for uniformity
from pyspark.sql.functions import upper, trim, col

df_clean = df_clean.withColumn("block", upper(df_clean["block"]))

# remove trailing and leading spaces
df_clean = df_clean.withColumn("block", trim(col("block")))
```

```python
# number of unique values for all categorical variables
categorical_cols = ["block", "iucr", "primary type", "description", "location description", "beat", "district", "ward", "community area", "fbi code"]

for col in categorical_cols:
    print(f"{col}: {df_clean.select(col).distinct().count()} unique values")
```

```python
# column entries that appear less than 5 times
for col in categorical_cols:
    print(f"\n{col} rare values (appearing <= 5 times):")
    df_clean.groupBy(col).count().filter("count <= 5").orderBy("count").show()
```

```python
# simple counts for the boolean vars
boolean_cols = ["arrest", "domestic"]

for col in boolean_cols:
    print(f"\n{col} value counts:")
    df_clean.groupBy(col).count().orderBy(col).show()
```

### Cleaning column 'block'

```python
# this cell looks at seeing if there are different ways to write street, such as ST, ST., STREET, etc.
# they all stick to one uniform way

from pyspark.sql.functions import col
# Count ending in " ST."
st_dot_count = df_clean.filter(col("block").endswith(" ST.")).count()
# Count ending in " ST"
st_count = df_clean.filter(col("block").endswith(" ST")).count()
# Count ending in " STREET"
street_count = df_clean.filter(col("block").endswith(" STREET")).count()


# Count ending in " AV"
av_count = df_clean.filter(col("block").endswith(" AV")).count()
# Count ending in " AVE"
ave_count = df_clean.filter(col("block").endswith(" AVE")).count()
# Count ending in " AVENUE"
avenue_count = df_clean.filter(col("block").endswith(" AVENUE")).count()

# Count ending in " ."
dot_count = df_clean.filter(col("block").endswith(".")).count()

print(f"Ends with 'ST.': {st_dot_count}")
print(f"Ends with 'ST' : {st_count}")
print(f"Ends with 'STREET': {street_count}")

print(f"Ends with 'AV': {av_count}")
print(f"Ends with 'AVE' : {ave_count}")
print(f"Ends with 'AVENUE': {avenue_count}")

print(f"Ends with '.' : {dot_count}")
```

```python
# finding how many blocks appear just once
rare_blocks = df_clean.groupBy("block").count().filter(col("count") == 1)
rare_rows_count = df_clean.join(rare_blocks, on="block", how="inner").count()

print(f"Rows with blocks that appear only once: {rare_rows_count}")
```

### Cleaning column 'primary type'

```python
# Examining 'primary type' values 
print(f"'primary type' values:")
df_clean.groupBy('primary type') \
    .count() \
    .orderBy("count") \
    .show(n = 40, truncate=False)
```

```python
# need to bring NON-CRIMINAL to one standard variable
# also grouping CRIM SEXUAL ASSAULT and CRIMINAL SEXUAL ASSAULT to one
from pyspark.sql.functions import when
from pyspark.sql import functions as F

df_clean = df_clean.withColumn(
    'primary type',
    F.when(F.col('primary type').isin('NON - CRIMINAL', 'NON-CRIMINAL (SUBJECT SPECIFIED)'), 'NON-CRIMINAL')
    .when(F.col('primary type').isin('CRIM SEXUAL ASSAULT'), 'CRIMINAL SEXUAL ASSAULT')
    .otherwise(F.col('primary type'))  
)
```

```python
# dropping non-criminal as only 182 rows, 
# this way we are focussing more on actual crimes
df_clean = df_clean.filter(df_clean['primary type'] != 'NON-CRIMINAL')
```

```python
df_clean = df_clean.withColumn(
    'primary type',
    F.when(F.col('primary type').isin('ASSAULT', 'BATTERY', 'HOMICIDE', 'ROBBERY', 'KIDNAPPING', 'INTIMIDATION', 'STALKING'), 'VIOLENT CRIME')

    .when(F.col('primary type').isin('SEX OFFENSE', 'PROSTITUTION', 'OBSCENITY', 'PUBLIC INDECENCY', 'CRIMINAL SEXUAL ASSAULT', 'HUMAN TRAFFICKING'), 'SEXUAL CRIME')

    .when(F.col('primary type').isin('PUBLIC PEACE VIOLATION', 'INTERFERENCE WITH PUBLIC OFFICER','LIQUOR LAW VIOLATION', 'CONCEALED CARRY LICENSE VIOLATION','WEAPONS VIOLATION', 'GAMBLING', 'RITUALISM'), 'PUBLIC ORDER')

    .when(F.col('primary type').isin('NARCOTICS', 'OTHER NARCOTIC VIOLATION'), 'DRUG RELATED')

    .when(F.col('primary type').isin('THEFT', 'BURGLARY', 'MOTOR VEHICLE THEFT', 'ARSON','CRIMINAL DAMAGE', 'CRIMINAL TRESPASS'), 'PROPERTY CRIME')

    .when(F.col('primary type').isin('DECEPTIVE PRACTICE', 'OFFENSE INVOLVING CHILDREN'), 'WHITE-COLLAR/REGULATORY')

    .otherwise(F.col('primary type'))
)
```

```python
# Examining 'primary type' rare values after cleaning
print(f"'primary type' values:")
df_clean.groupBy('primary type') \
    .count() \
    .orderBy("count") \
    .show(n = 40, truncate=False)
```

### Cleaning column 'location description'

```python
# remove trailing and leading spaces
df_clean = df_clean.withColumn("location description", trim(col("location description")))

df_clean.select('location description').distinct().count() # 201 originally
```

```python
# using a regex to remove spaces before and after "/" and "-" that are found in many of the descriptions.
from pyspark.sql.functions import regexp_replace, col

# removing spaces before and after "/" and "-"
df_clean = df_clean.withColumn(
    'location description',
    regexp_replace(col('location description'), r'\s*/\s*', '/')
).withColumn(
    'location description',
    regexp_replace(col('location description'), r'\s*-\s*', '-')
)
```
