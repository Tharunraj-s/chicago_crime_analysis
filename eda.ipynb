Analysis of crime incidents by Month, Hour of the day, Day of the week respectively - 

(In day of the week part, 1 = Sunday...7 = Saturday)

```
from pyspark.sql.functions import to_timestamp, month, hour, dayofweek

df = df.withColumn(
    "datetime",
    to_timestamp(col("date"), "dd/MM/yyyy HH:mm")
).withColumn("month", month(col("datetime"))) \
 .withColumn("hour", hour(col("datetime"))) \
 .withColumn("dow", dayofweek(col("datetime")))

display(df.groupBy("month").count().orderBy("month"))
display(df.groupBy("hour").count().orderBy("hour"))
display(df.groupBy("dow").count().orderBy("dow"))
```

```
from pyspark.sql.functions import hour, dayofweek, corr

df3 = (
    df.withColumn("ts", to_timestamp(col("date"), "dd/MM/yyyy HH:mm"))
      .withColumn("hr", hour(col("ts")))
      .withColumn("dow", dayofweek(col("ts")))
)

#Correlateing hour-of-day with day-of-week
df3.select(corr("hr", "dow")).show()
```

The correlation coefficient of 0.02 indicates that there is no general relationship between hour and day of week. This indicates that crimes do not occur on specific hours in a day.

```
coords = (
    df
      .select("latitude", "longitude")
      .sample(fraction=0.1)  #10% of the dataset
      .toPandas()
)
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 8))

plt.hexbin(
    coords["longitude"],
    coords["latitude"],
    gridsize=100,
    mincnt=1,     
    cmap="Reds"
)

plt.title("Chicago Crime Density Heatmap")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.colorbar(label="Crime Count")
plt.show()
```

```
from pyspark.sql.functions import to_timestamp, year, col
import matplotlib.pyplot as plt

df2 = (
    df
      .withColumn("ts", to_timestamp(col("date"), "dd/MM/yyyy HH:mm"))
      .withColumn("yr", year(col("ts")))
)

df2 = df2.filter(col("yr").isNotNull())
yearly_df = (
    df2
      .groupBy("yr")
      .count()
      .orderBy("yr")
)
yearly_pdf = yearly_df.toPandas()
yearly_pdf["yr"] = yearly_pdf["yr"].astype(int)

plt.figure(figsize=(10,6))
plt.plot(
    yearly_pdf["yr"],
    yearly_pdf["count"],
    marker="o",
    linestyle="-",
    linewidth=2
)
plt.xlabel("Year")
plt.ylabel("Number of Crimes")
plt.title("Crime Trend Over the Years")
plt.xticks(yearly_pdf["yr"], rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```

display(df.groupBy("arrest").count())#Arrest vs No-arrest
df2 = (
    df
    .withColumn("ts", to_timestamp(col("date"), "dd/MM/yyyy HH:mm"))
    .select("arrest", "latitude", "longitude")
    .dropna()
)
sample_pdf = df2.sample(fraction=0.01).toPandas()
sample_pdf["arrest_label"] = sample_pdf["arrest"].apply(lambda x: "Arrest" if x else "No Arrest")

plt.figure(figsize=(8, 8))
for label, group in sample_pdf.groupby("arrest_label"):
    plt.scatter(group["longitude"], group["latitude"], s=1, alpha=0.3, label=label)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Crime Locations Colored by Arrest Outcome")
plt.legend(markerscale=5)
plt.tight_layout()
plt.show()
```

```
df2 = (
    df
      .withColumn("ts", to_timestamp(col("date"), "dd/MM/yyyy HH:mm"))
      .withColumn("yr", year(col("ts")))
      .filter(col("yr").isNotNull())
)

agg = (
    df2
      .groupBy("yr", "primary type")
      .count()
)

trend_pdf = (
    agg
      .groupBy("yr")
      .pivot("primary type")
      .sum("count")
      .orderBy("yr")
      .toPandas()
      .set_index("yr")
)

plt.figure(figsize=(12, 6))
for crime in trend_pdf.columns:
    plt.plot(trend_pdf.index, trend_pdf[crime], marker="o", label=crime)

plt.title("Yearly Trends by Crime Category")
plt.xlabel("Year")
plt.ylabel("Number of Incidents")
plt.xticks(trend_pdf.index, rotation=45)
plt.legend(title="Primary Type", bbox_to_anchor=(1,1))
plt.tight_layout()
plt.show()
```

```
from pyspark.sql.functions import col, count

community_heat = (
    df
      .filter(col("Community Area").isNotNull())
      .groupBy("Community Area")
      .agg(count("*").alias("Total"))
      .orderBy("Community Area")
      .toPandas()
)
display(community_heat)
```

```
top10 = community_heat.sort_values("Total", ascending=False).head(10)
labels = top10["Community Area"].astype(str)
sizes  = top10["Total"]

plt.figure(figsize=(8,8))
plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Crime Distribution Among Top 10 Community Areas")
plt.axis("equal")  
plt.tight_layout()
plt.show()
```

Community area 25 reported the maximum number of crimes.

```
from pyspark.sql.functions import count

ward_counts = (
    df
      .groupBy("Ward")
      .agg(count("*").alias("CrimeCount"))
      .orderBy("Ward")
)

ward_pd = ward_counts.toPandas()
ward_pd["Ward"] = ward_pd["Ward"].astype(str)
plt.figure(figsize=(10, 8))
plt.barh(
    ward_pd["Ward"],
    ward_pd["CrimeCount"]
)
plt.xlabel("Crime Count")
plt.ylabel("Ward")
plt.title("Crime Count by Ward")
plt.tight_layout()
plt.show()
```
